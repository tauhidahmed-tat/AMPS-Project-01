```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Amazon_BestSelling_Books_500.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
x=df[['Rating','Reviews']].to_numpy()
shape=x.shape
size=x.size
ndim=x.ndim
dtype=x.dtype

print(shape)
print(size)
print(ndim)
print(dtype)
```

```python
print(np.sum(x))
print(np.mean(x))
print(np.median(x))
print(np.std(x))
print(np.min(x))
print(np.max(x))
print(np.percentile(x,25))
print(np.percentile(x,50))
print(np.percentile(x,75))
```

```python
print("\n---Column-wise(axis=0)---")
print(np.sum(x,axis=0))
print(np.mean(x,axis=0))
print(np.median(x,axis=0))
print(np.std(x,axis=0))
print(np.min(x,axis=0))
print(np.max(x,axis=0))
print(np.percentile(x,25,axis=0))
print(np.percentile(x,50,axis=0))
print(np.percentile(x,75,axis=0))
```

```python
print("----Row-wise(axis=1)----")
print(np.sum(x,axis=1))
print(np.mean(x,axis=1))
print(np.median(x,axis=1))
print(np.std(x,axis=1))
print(np.min(x,axis=1))
print(np.max(x,axis=1))
print(np.percentile(x,25,axis=1))
print(np.percentile(x,50,axis=1))
print(np.percentile(x,75 ,axis=1))
```
